```python
df['calculated_sales'] = (
    df['quantity'] * df['unit_price'] * (1 - df['discount_pct'] / 100)
)
```

```python
(df['sales_amount'] - df['calculated_sales']).abs().describe()
```

```python
df.drop(columns='calculated_sales', inplace=True)
```

Conclusão – Compreensão e Qualidade dos Dados

A análise inicial mostrou que o dataset contém 120.000 transações e 17 variáveis, abrangendo o período de janeiro de 2024 a dezembro de 2025. Não foram identificados valores ausentes, registros duplicados ou IDs de transação repetidos. A variável transaction_date foi convertida para o formato datetime, e as variáveis numéricas apresentaram valores dentro de intervalos plausíveis. Além disso, o sales_amount foi validado com base na quantidade, preço unitário e desconto, apresentando apenas pequenas diferenças decorrentes de arredondamentos.

A análise das variáveis categóricas identificou 8 categorias de produtos, 40 produtos, 3 marcas e 3 canais de venda, com uma distribuição bastante equilibrada entre as categorias e canais. Os segmentos de clientes e grupos etários também apresentam uma distribuição semelhante entre si.

De modo geral, os dados apresentam boa qualidade e estão estruturados de forma consistente, sendo adequados para avançar para a Análise Exploratória de Dados (EDA).
